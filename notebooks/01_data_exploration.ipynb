### Notebook Objectives
1. Ingest and profile the ESCO technical skills taxonomy dataset (`skills_taxonomy.csv`).
2. Analyze skill distribution across technical domains (Machine Learning, DevOps, Cloud, etc.).
3. Examine predefined tech job profiles and required skill density.
4. Analyze skill co-occurrence and industry demand distributions.

```python
import os
import csv
from collections import Counter
import matplotlib.pyplot as plt

# 1. Load Skills Taxonomy
taxonomy_path = '../dataset/skills/skills_taxonomy.csv'
skills = []
with open(taxonomy_path, 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        skills.append(row)

print(f'Total skills loaded: {len(skills)}')
print('Sample record:', skills[0])
```

### Data Inspection
The taxonomy contains standardized entries with preferred labels, alternative names/acronyms, assigned technical categories, and knowledge/skill classifications. This forms our canonical knowledge base.

```python
# 2. Analyze Category Breakdown
categories = [s['category'] for s in skills]
cat_counts = Counter(categories)
for cat, count in cat_counts.most_common():
    print(f'{cat:25s}: {count:3d} skills ({count/len(skills)*100:.1f}%)')
```

### Category Distribution Findings
The taxonomy spans 10 distinct technical domains, ensuring balanced coverage across ML/AI, Data Science, Data Engineering, Backend, Frontend, Cloud, and DevOps disciplines.

```python
# 3. Visualize Category Distribution
sorted_cats = sorted(cat_counts.items(), key=lambda x: x[1], reverse=True)
names = [x[0] for x in sorted_cats]
counts = [x[1] for x in sorted_cats]

plt.figure(figsize=(10, 5))
plt.barh(names[::-1], counts[::-1], color='#3B82F6')
plt.title('ESCO Skills Taxonomy: Domain Distribution')
plt.xlabel('Number of Canonical Skills')
plt.tight_layout()
plt.show()
```

### Plot Interpretation
The distribution indicates high representation in Machine Learning, Programming, and Data Science, which directly aligns with the core requirements of tech industry hiring workflows.

```python
# 4. Explore Job Roles Catalog
jobs_path = '../dataset/jobs/job_roles.csv'
jobs = []
with open(jobs_path, 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        reqs = [s.strip() for s in row['required_skills'].split('|') if s.strip()]
        jobs.append({
            'role_name': row['role_name'],
            'req_count': len(reqs),
            'req_skills': reqs
        })

for j in jobs:
    print(f"{j['role_name']:30s} -> {j['req_count']} required skills")
```

### Job Role Requirements Analysis
Target tech roles demand between 7 and 9 core competencies on average, establishing a realistic requirement threshold for gap analysis.

```python
# 5. Top In-Demand Skills Across Job Profiles
all_req_skills = [s for j in jobs for s in j['req_skills']]
skill_demand = Counter(all_req_skills)

top_skills = skill_demand.most_common(10)
s_names = [x[0] for x in top_skills]
s_counts = [x[1] for x in top_skills]

plt.figure(figsize=(9, 4))
plt.bar(s_names, s_counts, color='#10B981')
plt.title('Top Most Frequently Demanded Skills Across Tech Roles')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Role Frequency Count')
plt.tight_layout()
plt.show()
```

### Top Demanded Skills Interpretation
Foundational skills such as `Python`, `SQL`, `Git`, and `Docker` cross multiple job role categories, highlighting the necessity of multi-role skill resolution.
